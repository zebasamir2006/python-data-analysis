```python
import pandas as pd

url= "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df=pd.read_csv(url)
df.head()

import seaborn as sns
import matplotlib.pyplot as plt
sns.scatterplot(x="Age",y="Fare",data=df)
plt.title("age vs fare")
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
df["Embarked"].fillna("S",inplace=True)
df['Age'].fillna(df['Age'].mean(),inplace=True)
```

```python
df['deck'] = df['Cabin'].str[0]
df['deck'].fillna('Unknown', inplace=True)
```

```python
#dropped class,who,adult_male,deck,embark_town,alive
```

```python
df["Sex"]=df["Sex"].map({"male":0,"female":1})
df["Embarked"]=df["Embarked"].map({"S":0,"C":1,"Q":2})
df['alone'] = ((df['SibSp'] + df['Parch']) == 0)
df["alone"]=df["alone"].map({True:1,False:0})
```

```python
x=df.drop(["Survived", "Name", "Ticket", "Cabin", "deck", "PassengerId"], axis=1)
y=df["Survived"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(
    x,y,test_size=0.2,random_state=42
)
```

```python
model=LogisticRegression(max_iter=1000)
model.fit(x_train,y_train)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
x_train['column_name'] = le.fit_transform(x_train['column_name'])
```

```python
print(x_train.dtypes)
```

```python
x_train = x_train.drop(columns=['Name', 'Ticket', 'Cabin'])
x_test = x_test.drop(columns=['Name', 'Ticket', 'Cabin'])
```

```python
x_train = pd.get_dummies(x_train)
x_test = pd.get_dummies(x_test)
```

```python
x_train, x_test = x_train.align(x_test, join='left', axis=1, fill_value=0)
```

```python
model.fit(x_train, y_train)
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy='mean')

x_train = imputer.fit_transform(x_train)
x_test = imputer.transform(x_test)
```

```python
model.fit(x_train, y_train)
```

```python
y=pred=model.predict(x_test)
```

```python
# Train
model.fit(x_train, y_train)

# Predict
y_pred = model.predict(x_test)

# Accuracy
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test, y_pred))
```

```python
y_pred = model.predict(x_test)
```

```python
print(accuracy_score(y_test, y_pred))
```

```python
cm=confusion_matrix(y_test,y_pred)
sns.heatmap(cm,annot=True,fmt="d")
plt.title("confusion matrix")
plt.show()
```
